```python
raw = pd.read_csv(path + 'labeled_data.csv')
print(raw.columns.tolist())
raw.head()
```

```python
print(raw.dtypes.to_string())
```

```python
for c in raw.select_dtypes('object').columns:
    print(c, '|', raw[c].nunique(), '종류 |', raw[c].unique()[:5])
```

```python
print('전체 중복 행:', raw.duplicated().sum())
print('_id 중복:', raw['_id'].duplicated().sum())
display(pd.crosstab(raw['PART_NAME'], raw['PassOrFail'], margins=True))
display(pd.crosstab(raw['PART_NAME'], raw['Reason'].fillna('-')))
display(pd.crosstab(raw['PART_NAME'], raw['EQUIP_NAME']))
```

```python
df0 = raw.drop_duplicates().copy()
print('중복 제거 후:', df0.shape)

cn7 = df0[df0['PART_NAME'].str.contains('CN7')].copy()
display(pd.crosstab(cn7['PART_NAME'], cn7['Reason'].fillna('양품'), margins=True))
print('CN7 행 수:', len(cn7), '/ 고유 시각 수:', cn7['TimeStamp'].nunique())
```

```python
t = cn7['TimeStamp'].iloc[0]
cn7[cn7['TimeStamp'] == t].T
```

```python
vc = cn7['TimeStamp'].value_counts()
print(vc.value_counts())
t2 = vc[vc == 2].index[0]
pair = cn7[cn7['TimeStamp'] == t2]
num = pair.select_dtypes('number').columns.drop('PART_FACT_SERIAL')
print('좌우 공정값이 모두 같은가:', (pair[num].nunique() == 1).all())
pair[['PART_NAME', 'PassOrFail', 'Reason']]
```

```python
nums = cn7.select_dtypes('number')
print('상수 컬럼:', nums.columns[nums.nunique() <= 1].tolist())
print('스크류 RPM Max<Average 비율:', (cn7['Max_Screw_RPM'] < cn7['Average_Screw_RPM']).mean())
print('배압 Max<Average 비율:', (cn7['Max_Back_Pressure'] < cn7['Average_Back_Pressure']).mean())
```

```python
drop_cols = ['_id', 'PART_FACT_PLAN_DATE', 'PART_FACT_SERIAL', 'EQUIP_CD', 'EQUIP_NAME',
             'Switch_Over_Position', 'Barrel_Temperature_7'] + \
            [f'Mold_Temperature_{i}' for i in [1, 2, 5, 6, 7, 8, 9, 10, 11, 12]]

cn7['TimeStamp'] = pd.to_datetime(cn7['TimeStamp'])
cn7['fail'] = (cn7['PassOrFail'] == 'N').astype(int)
proc = [c for c in cn7.select_dtypes('number').columns if c not in drop_cols + ['fail']]

shot = cn7.groupby('TimeStamp').agg(
    **{c: (c, 'first') for c in proc},
    fail=('fail', 'max'),
    reason=('Reason', lambda s: ','.join(sorted(s.dropna().unique())))
).reset_index().sort_values('TimeStamp')

# 가정: 99.5% 행에서 Max < Average → 두 컬럼 이름이 바뀐 것으로 판단
shot = shot.rename(columns={'Max_Back_Pressure': 'Average_Back_Pressure',
                            'Average_Back_Pressure': 'Max_Back_Pressure'})

print(shot.shape, '/ 불량 샷:', shot['fail'].sum())
print(shot['reason'].value_counts())
shot.to_csv(path + 'cn7_shot_clean.csv', index=False)
```
